creation d'un programme qui demande nom prenom age son statut puis affiche une phrase qui affiche ses informations.
demander le nom
demande le prenom
demander l'age
demander le statut
puis afficher les infos

```python
def script1():                                                                    #definition d'une fonction
    nom=input("entrer votre nom")                                                 #on demande le nom
    prenom=input("entrer votre prenom")                                           #demande le prenom
    age=input("quel age a tu")                                                    #demande age
    statut=input("quel est ton statut étudiant ou autre")                         #demande du statut
    print("nom " + nom+" prenom "+prenom+" age "+age+" statut "+statut)           #affiche les infos entrer

script1()
```

definir le Solde
definir un depot
definir un retrait
demander l'operation effectuer
Mettre a jour le Solde
crer une alerte si solde dans le négatif

```python
def script4():
    S=0
    while S==0:
        input("quel operation voulez vous effectuer")
        D=int(input("combien souhaiter vous deposer ?"))
        S=D+S
        R=int(input("combien souhaiter vous retirer ?"))
        S=S-R
    print(S)
    if S<0:
          print ("Attention votre Solde est négatif")
script4()
```
